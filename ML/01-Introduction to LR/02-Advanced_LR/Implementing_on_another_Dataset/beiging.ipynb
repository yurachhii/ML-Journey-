```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('challenge-1-beijing.csv')
```

```python
df.head(10)
```

- **the features**(chinese characters) we choose include public transport, office, hospital, shopping mall, subway, school, construction time, floor and area - total nine items. We note that 
- per-square-meter price is **the target value**.

```python
features = df[df.columns[[0,1 ,2,3, 4,5,7,9,11]]]
features.head(10)
```

```python
target = df.iloc[: ,10]
target.head(10)
```

```python
pd.concat([features, target], axis=1).head()
```

```python
x_train , x_test , y_train , y_test = train_test_split(features , target , test_size = 0.3 , random_state = 42)
```

```python
len(x_train), len(x_test), len(y_train), len(y_test)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)

intercept = model.intercept_
coefficients = model.coef_ 
```

- we introduce another evaluation indicator, which is average absolute percentage error **MAPE**
- **MAPE** is a variant of MAE, which is a percentage value and, therefore, is easier to understand than other statistics

```python
def mape(y_true, y_pred):
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100
```

```python
y_true = y_test.values
y_pred = model.predict(x_test)
mape(y_true, y_pred)
```
