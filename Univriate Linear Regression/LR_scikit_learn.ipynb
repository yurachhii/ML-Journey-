```python
from sklearn.linear_model import LinearRegression
```

```python
import numpy as np
x = np.array([56, 72, 69, 88, 102, 86, 76, 79, 94, 74])
y = np.array([92, 102, 86, 110, 130, 99, 96, 102, 105, 92])
```

```python
model = LinearRegression() 
model.fit(x.reshape(len(x), 1), y) # it is important to reshape the x array to be a 2D array with shape (n_samples, n_features) for scikit-learn's LinearRegression model.
model.coef_, model.intercept_ # the slope and intercept of the linear regression line
```

```python
print(model.predict([[150]]))
```
