```python
import numpy as np
# first using univriate linear regression 
x = np.array([56, 72, 69, 88, 102, 86, 76, 79, 94, 74])
y = np.array([92, 102, 86, 110, 130, 99, 96, 102, 105, 92])
```

```python
# drawing a scatter plot
plt.scatter(x, y)
plt.xlabel("Area")
plt.ylabel("Price")
```

```python
# building univariate linear regression model
def f(x, b, w):
    y_hat = b + w * x
    return y_hat
```

```python
# start to calculte the loss function
def square_function(y,x,b,w):
    y_hat = f(x,b,w)
    loss = sum(np.square(y-y_hat))/(len(y)*2)
    return loss
```

```python
def w_and_b(x,y):
    m = len(x)
    upper = m * sum(x*y) - sum(x)*sum(y)
    lower = m * sum(x**2) - (sum(x))**2
    w = upper / lower 
    b = (sum(y) - w * sum(x)) / m
    return w, b
```

```python
w_, b_ = w_and_b(x,y)
print("w: ", w_)
print("b: ", b_)
```

```python
print(square_function(y,x,b_,w_))
```

```python
x_test = np.array([50,120,100])
plt.scatter(x,y)
plt.plot(x_test, f(x_test, b_, w_))
```
