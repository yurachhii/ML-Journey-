```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('Boston.csv')
```

```python
df.head()
```

```python
df.duplicated().sum()
```

```python
# we will just be using as inputs
# 1-  Urban crime rate (CRIM) 
# 2-  The average number of rooms in the home (RM)
# 3-  Percentage of people with lower status in the population (LSTAT)
# ----------------------------------------------------------------
# 4- Median value of owner-occupied homes in $1000's (MEDV) - this is the target variable
features = df[['crim','rm','lstat']] 
target = df['medv']
```

```python
x_train , x_test , y_train , y_test = sklearn.model_selection.train_test_split(features,target,test_size=0.3,random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train,y_train) 
model.coef_, model.intercept_

```

```python
predictions = model.predict(x_test) 
predictions
```

```python
def msq_value(y_test, predictions):
    n = len(y_test)
    return np.sum(np.square(y_test - predictions)) / n 
```

```python
print("Mean Squared Error:", msq_value(y_test, predictions))
```
