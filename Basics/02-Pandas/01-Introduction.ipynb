**Why Use Pandas?**
- Pandas allows us to analyze big data and make conclusions based on statistical theories.

- Pandas can clean messy data sets and make them readable and relevant.

- Relevant data is very important in data science.

**What Pandas Can Do?**
- Pandas determine if there any correlation between 2 or more coulmns .

- Detrmine the Min , Max and Mean value . 

```python
import pandas as pd

dataset = {
    'cars' : ["BMW", "Volvo", "Ford"],
  'passings': [3, 7, 2]
}

df = pd.DataFrame(dataset)

print(df)
```

**What Is Series?**

- It is a one-dimensional array holding data of any type.

```python
row = [1 , 2 , 3]

var = pd.Series(row)

print(var)
```

**Labels**
- if you did not specify any indexing so the default is 0-indexing

- you can specify any kind of indexing at the 2nd parameter


```python
var1 = pd.Series(row, ['1st','2nd','3rd'] )

print(f"the 1st value is {var[0]} and the 2nd value is {var1['2nd']}")
```

**Data Frames**

- dataframe is 2 dimensional array -> table contains rows and coulmns 

```python
df = {
    'Calories' :  [150 , 200 , 400 , 470] ,
    'Duration' : [30 , 40 , 50 , 60]
}

df = pd.DataFrame(df)

print(df)
```

**Locate Row**

- Pandas use the loc attribute to return one or more specified row

- using **.loc** to DataFrame with indexing , range , specific valudes

```python
print(df.loc[0:2])
```

**Indexing**

-  as we have mentioned before we can totally use 0-indexing or specify it as we want , just differ as syntax but the exact concept

```python
data = {
    'Characters' : ['Ang' , 'Ketara' , 'Zoko' , 'Though'] , 
    'Bending' : ['Air-Bender' , 'Water-Bender' , 'Fire-Bender' , 'Earth-Bender']
}

df = pd.DataFrame(data , index = ['1st' , '2nd' , '3rd' , '4th'])

print(df.loc['1st':'3rd'])
```

    Viewing the Data
 -   One of the most used method for getting a quick overview of the DataFrame, is the head() method.

-    The head() method returns the headers and a specified number of rows, starting from the top

     Note: if the number of rows is not specified, the head() method will return the top 5 rows.

-    There is also a tail() method for viewing the last rows of the DataFrame.

-    The tail() method returns the headers and a specified number of rows, starting from the bottom.

```python
print(df1.head(10))
```

```python
print(df1.tail())
```
