```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data (1).csv')

mean = df['Calories'].mean()
df.fillna({'Calories': mean} , inplace = True)

# df['Date'] = pd.to_datetime(df['Date'] , format = 'mixed') 


for i in df.index:
    if df.loc[i, 'Duration'] > 120 :
        df.loc[i, 'Duration'] =120

df.drop_duplicates(inplace = True)
```

**Finding Relationships**

corr() method -> calculate the relationships between every coulmn in your dataset 

correlation determines the bond between 2 coulmns is it strong or weak 

and what happen if one coulmn increase what will happen to the other one 

the result must stay between -1 and 1

```python
df.corr()
```

```python
df.plot()
plt.show()
```

```python
# using scatter plot -> u have to specify the kind , x-axis and y-axis
df.plot(kind = 'scatter' , x = 'Duration' , y = 'Calories') # as we saw the correlation between those were 0.9 so the scatter plot shows the strong relationship
#Maxpulse
```
