```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Draw a line in a diagram from position (0,0) to position (6,250):

x = np.array([0,6])
y = np.array([0,250])

plt.plot(x,y)
plt.show()
```

**Plotting x and y points**

The plot() function is used to draw points (markers) in a diagram 

te default of the plot() will draw a simple line 

 - Parameter 1 is an array containing the points on the x-axis

 - Parameter 2 is an array containing the points on the y-axis



if u wanna plot points not a line u will add parameter **'o'**

```python
plt.plot(x,y,'o')

plt.show()
```

You can plot as many points as you like, just make sure you have the same number of points in both axis 

```python
x_points = np.array([1,2,6,8])
y_points = np.array([3,8,1,10])

plt.plot(x_points , y_points)

plt.show()
```

**Deafult X points**

it u didn't specify the x-points then the default value of them will be 0,1,2,3 and so on

```python
plt.plot(y_points)

plt.show()
```
