**Linestyle**

u can use argument linestyle or shorcut ls to change the style of the plotted line

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
y_points = np.array([3,10,5,9])

plt.plot(y_points , ls = 'dotted')

plt.show()
```

```python
plt.plot(y_points , ls = 'dashed')

plt.show()
```

**Line Color**

u can use keyword argument **color** with shortcut **c**



```python
plt.plot(y_points , ls = 'dashed' , c = 'g')

plt.show()
```

**Line Width**

u can use keyword argument linewidth -> **lw** to change the width of the line

```python
plt.plot(y_points  , lw = 5)

plt.show()
```

**Multiple lines**

* u can add multiple lines of **plt.plot()**

* You can also plot many lines by adding the points for the **x- and y-axis** for each line in the same plt.plot() function


```python
x1 = np.array([0, 1, 2, 3])
y1 = np.array([3, 8, 1, 10])
x2 = np.array([0, 1, 2, 3])
y2 = np.array([6, 2, 7, 11])

plt.plot(x1, y1, x2, y2)
plt.show()
```
