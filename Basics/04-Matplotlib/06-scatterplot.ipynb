```python
import matplotlib.pyplot as plt
import numpy as np
```

**Scatter Plot**

**scatter()** function plot one dot of each observation , it needs 2 arrays with the same size 


```python
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])

plt.scatter(x, y , c = 'r' )
plt.show()
```

**Size of The dot**

u can change the size of dos by using **s** argument 



```python
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])

plt.scatter(x , y  , s = 50)
```

**Alpha Argument**

You can adjust the transparency of the dots with the **alpha** argument

```python
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])

plt.scatter( x, y , s = 50 , alpha = 0.7)

```
