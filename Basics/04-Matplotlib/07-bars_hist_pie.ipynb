```python
import matplotlib.pyplot as plt
import numpy as np
```

**Bar Chart**

* u can use **bar()** function to draw a normal bar graph
* u can use **barh()** function to draw a horizontal bar graph 

```python
x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.bar(x,y , width = 0.09)
```

```python
x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.barh(x,y , height = 0.09)
```

**Histogram**

histogram is showing frequancy disribution 

we use **hist()** function to create histogram

```python
x = np.random.normal(170, 10, 250) 

plt.hist(x)
```

**Creating Pie Chart**

- using **pie()** function to draw pie chart

- using **labels** parameter to define label of each wedge

- using **startangle** parameter to defined an angle in degrees, default angle is 0



```python

y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]

plt.pie(y , labels = mylabels , startangle = 270)
```

**Explode**

**explode** parameter will allow u to stand a wedge out  

```python

y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
myexplode = [0.2, 0, 0, 0]

plt.pie(y , labels = mylabels , explode = myexplode )
```

- to add shadow set **shadow** argument to True

- to add color of each wedget make a list of desirble colors then set it to **color** argument

- to add a list of explanation for each wedge use the legend() function -> to make it labeled use **title** argument

```python
y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
myexplode = [0.2, 0, 0, 0]
clr = ['r' , 'y' , 'm' , 'b']
plt.pie(y , labels = mylabels , explode = myexplode , shadow = True , colors= clr)
plt.legend(title = 'FRUITS')
plt.show()
```
